```python
import pandas as pd
import numpy as np
import re
import nltk
import joblib

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score
from sklearn.pipeline import Pipeline
```

```python
df = pd.read_csv("fake reviews dataset.csv.zip")

print(df.head())
print(df.info())
```

```python
df = df[['text_', 'label']]

df.dropna(inplace=True)
df.drop_duplicates(inplace=True)
```

```python
df['label'] = df['label'].map({
    'CG': 1,   # Fake Review
    'OR': 0    # Real Review
})

```

```python
print(df['label'].value_counts())
```

```python
df['review_length'] = df['text_'].apply(len)

print(df.groupby('label')['review_length'].mean())
```

```python
nltk.download('stopwords')
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
```

```python
def clean_text(text):

    # Lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r'http\S+', '', text)

    # Remove special characters
    text = re.sub(r'[^a-zA-Z ]', '', text)

    # Remove stopwords
    text = " ".join([
        word for word in text.split()
        if word not in stop_words
    ])

    return text
```

```python
df['clean_text'] = df['text_'].apply(clean_text)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df['clean_text'],
    df['label'],
    test_size=0.2,
    random_state=42
)
```

```python
tfidf = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1,2)
)
```

```python
pipeline = Pipeline([
    ('tfidf', tfidf),
    ('model', LogisticRegression())
])
```

```python
pipeline.fit(X_train, y_train)
```

```python
y_pred = pipeline.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))

print(classification_report(y_test, y_pred))
```

```python
joblib.dump(pipeline, "fake_review_model.pkl")
```

```python
model = joblib.load("fake_review_model.pkl")
```

```python
def predict_review(review):

    review = clean_text(review)

    prediction = model.predict([review])[0]

    if prediction == 1:
        return "Fake Review"
    else:
        return "Real Review"
```

```python
%%writefile app.py
import streamlit as st
import joblib
import re
import pandas as pd
from nltk.corpus import stopwords

# ---------------- UI CONFIG ----------------
st.set_page_config(
    page_title="Fake Review Detector",
    page_icon="🕵️",
    layout="wide"
)

# ---------------- STYLING ----------------
st.markdown("""
<style>
.big-title {
    font-size:40px;
    font-weight:bold;
    text-align:center;
    color:#FF4B4B;
}
.result-box {
    padding:20px;
    border-radius:10px;
    text-align:center;
    font-size:22px;
}
</style>
""", unsafe_allow_html=True)

# ---------------- LOAD MODEL ----------------
@st.cache_resource
def load_model():
    return joblib.load("fake_review_model.pkl")

model = load_model()

stop_words = set(stopwords.words('english'))

# ---------------- TEXT CLEANING ----------------
def clean_text(text):
    text = text.lower()
    text = re.sub(r'http\S+', '', text)
    text = re.sub(r'[^a-zA-Z ]', '', text)
    text = " ".join([w for w in text.split() if w not in stop_words])
    return text

# ---------------- PREDICTION ----------------
def predict_review(review):

    cleaned = clean_text(review)
    prediction = model.predict([cleaned])[0]

    if hasattr(model, "predict_proba"):
        prob = model.predict_proba([cleaned])[0]
    else:
        prob = [0.5, 0.5]

    return prediction, prob

# ---------------- SIDEBAR ----------------
st.sidebar.title("📊 Model Information")

st.sidebar.write("""
**Algorithm:** Logistic Regression
**Feature:** TF-IDF
**Task:** Fake Review Detection
""")

st.sidebar.markdown("---")
st.sidebar.write("💡 Paste review text to detect fraud")

# ---------------- MAIN TITLE ----------------
st.markdown('<p class="big-title">🕵️ Fake Review Detection System</p>', unsafe_allow_html=True)

st.write("Detect whether product reviews are **Fake or Real** using Machine Learning.")

# ---------------- INPUT SECTION ----------------
col1, col2 = st.columns([3,1])

with col1:
    review = st.text_area("✍️ Enter Product Review", height=150)

with col2:
    st.write("")
    st.write("")
    predict_btn = st.button("🔍 Predict")

# ---------------- PREDICTION OUTPUT ----------------
if predict_btn:

    if review.strip():

        pred, prob = predict_review(review)

        fake_prob = prob[1]
        real_prob = prob[0]

        # Result Display
        if pred == 1:
            st.error("🚨 Fake Review Detected")
            confidence = fake_prob
        else:
            st.success("✅ Real Review")
            confidence = real_prob

        # Confidence Progress Bar
        st.subheader("Confidence Score")
        st.progress(float(confidence))

        st.write(f"Confidence: {round(confidence*100,2)} %")

        # Probability Chart
        st.subheader("Prediction Probability")

        chart_data = pd.DataFrame({
            "Category": ["Real", "Fake"],
            "Probability": [real_prob, fake_prob]
        })

        st.bar_chart(chart_data.set_index("Category"))

    else:
        st.warning("⚠️ Please enter review text")

# ---------------- FOOTER ----------------
st.markdown("---")
st.caption("Built with ❤️ using NLP & Machine Learning")
```
